```python
"""
YC Winter 2026 Founder Scraper for Google Colab
Run this in a Colab notebook cell
"""

# Install Chrome and dependencies
print("Installing Chrome and dependencies...")
!wget -q https://dl.google.com/linux/direct/google-chrome-stable_current_amd64.deb
!apt-get install -y -qq ./google-chrome-stable_current_amd64.deb
!pip install selenium webdriver-manager -q

import sys
import time
import csv
import os
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, NoSuchElementException
from webdriver_manager.chrome import ChromeDriverManager
from google.colab import files
import pandas as pd

def setup_driver():
    """Initialize Chrome driver with Colab-compatible options"""
    chrome_options = webdriver.ChromeOptions()
    chrome_options.add_argument('--headless=new')
    chrome_options.add_argument('--no-sandbox')
    chrome_options.add_argument('--disable-dev-shm-usage')
    chrome_options.add_argument('--disable-gpu')
    chrome_options.add_argument('--disable-software-rasterizer')
    chrome_options.add_argument('--disable-extensions')
    chrome_options.add_argument('--disable-blink-features=AutomationControlled')
    chrome_options.add_argument('--window-size=1920,1080')
    chrome_options.add_argument('--remote-debugging-port=9222')
    chrome_options.add_argument('user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36')

    # Use webdriver_manager to handle ChromeDriver automatically
    service = Service(ChromeDriverManager().install())

    driver = webdriver.Chrome(service=service, options=chrome_options)
    driver.set_page_load_timeout(30)

    return driver

def get_company_links(driver, batch_url):
    """Extract all company links from the batch page"""
    print("Loading batch page...")
    driver.get(batch_url)
    time.sleep(4)

    company_links = []
    try:
        # Find all company links
        companies = driver.find_elements(By.CSS_SELECTOR, 'a[href*="/companies/"]')

        for company in companies:
            href = company.get_attribute('href')
            if href and '/companies/' in href and '?' not in href and href not in company_links:
                company_links.append(href)

        # Remove duplicates and filter out the batch page itself
        company_links = list(set([link for link in company_links if not link.endswith('/companies')]))

        print(f"✓ Found {len(company_links)} companies")
        return company_links

    except Exception as e:
        print(f"✗ Error getting company links: {e}")
        return []

def get_founder_details(driver, company_url):
    """Extract founder details from a company page"""
    company_name = company_url.split('/')[-1]
    print(f"  Processing: {company_name}...", end=" ")

    try:
        driver.get(company_url)
        time.sleep(2)

        founders_data = []

        # Wait for page to load
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.TAG_NAME, "body"))
        )

        # Strategy: Collect all social links from the page
        try:
            all_links = driver.find_elements(By.TAG_NAME, 'a')

            twitter_links = []
            linkedin_links = []

            for link in all_links:
                href = link.get_attribute('href')
                if href:
                    if ('twitter.com' in href or 'x.com' in href) and '/i/' not in href:
                        twitter_links.append(href)
                    elif 'linkedin.com/in/' in href:
                        linkedin_links.append(href)

            # Remove duplicates while preserving order
            twitter_links = list(dict.fromkeys(twitter_links))
            linkedin_links = list(dict.fromkeys(linkedin_links))

            # Try to find founder names
            founder_names = []

            # Look for text near "Founder" or "Co-founder" labels
            text_elements = driver.find_elements(By.XPATH,
                "//*[contains(text(), 'Founder') or contains(text(), 'founder')]/..//*[self::h1 or self::h2 or self::h3 or self::h4 or self::p or self::span or self::div]")

            for elem in text_elements:
                text = elem.text.strip()
                # Filter for names (typically 2-4 words, starts with capital)
                words = text.split()
                if text and 2 <= len(words) <= 4 and text[0].isupper() and 'Founder' not in text and 'founder' not in text:
                    if text not in founder_names:
                        founder_names.append(text)

            # Create founder entries
            max_founders = max(len(twitter_links), len(linkedin_links), 1)

            for i in range(max_founders):
                founder_data = {
                    'company': company_name,
                    'company_url': company_url,
                    'founder_name': founder_names[i] if i < len(founder_names) else f'Founder {i+1}',
                    'twitter': twitter_links[i] if i < len(twitter_links) else 'N/A',
                    'linkedin': linkedin_links[i] if i < len(linkedin_links) else 'N/A'
                }
                founders_data.append(founder_data)

            if founders_data:
                print(f"✓ ({len(founders_data)} founders)")
            else:
                print("⚠ No founders found")
                founders_data.append({
                    'company': company_name,
                    'company_url': company_url,
                    'founder_name': 'N/A',
                    'twitter': 'N/A',
                    'linkedin': 'N/A'
                })

        except Exception as e:
            print(f"⚠ Partial data")
            founders_data.append({
                'company': company_name,
                'company_url': company_url,
                'founder_name': 'N/A',
                'twitter': 'N/A',
                'linkedin': 'N/A'
            })

        return founders_data

    except Exception as e:
        print(f"✗ Error: {str(e)[:50]}")
        return [{
            'company': company_name,
            'company_url': company_url,
            'founder_name': 'Error',
            'twitter': 'Error',
            'linkedin': 'Error'
        }]

def main():
    batch_url = "https://www.ycombinator.com/companies?batch=Winter%202026"
    output_file = "yc_winter_2026_founders.csv"

    print("="*60)
    print("YC Winter 2026 Founder Scraper")
    print("="*60)

    driver = None

    try:
        print("Initializing Chrome driver...")
        driver = setup_driver()
        print("✓ Chrome driver initialized successfully\n")

        all_founders = []

        # Step 1: Get all company links
        company_links = get_company_links(driver, batch_url)

        if not company_links:
            print("\n✗ No companies found. The page structure may have changed.")
            return None

        print(f"\nStarting to scrape {len(company_links)} companies...")
        print("-"*60)

        # Step 2 & 3: Process each company
        for idx, company_url in enumerate(company_links, 1):
            print(f"[{idx}/{len(company_links)}]", end=" ")
            founders = get_founder_details(driver, company_url)
            all_founders.extend(founders)

            # Show progress every 10 companies
            if idx % 10 == 0:
                print(f"\n  Progress: {idx}/{len(company_links)} companies processed\n")

            time.sleep(2)  # Be respectful with requests

        # Save to CSV
        if all_founders:
            # Create DataFrame for better display
            df = pd.DataFrame(all_founders)

            # Save to CSV
            df.to_csv(output_file, index=False)

            print("\n" + "="*60)
            print("✓ SCRAPING COMPLETE!")
            print("="*60)
            print(f"✓ Total companies processed: {len(company_links)}")
            print(f"✓ Total founder entries: {len(all_founders)}")
            print(f"✓ Data saved to: {output_file}")

            # Display preview
            print("\n📊 Preview of data:")
            print(df.head(10).to_string())

            # Download file
            print(f"\n⬇️  Downloading {output_file}...")
            files.download(output_file)

            return df
        else:
            print("\n✗ No data collected.")
            return None

    except Exception as e:
        print(f"\n✗ Error in main execution: {e}")
        import traceback
        traceback.print_exc()
        return None

    finally:
        if driver:
            driver.quit()
            print("\n✓ Browser closed.")

# Run the scraper
print("Starting scraper...\n")
df = main()
```
